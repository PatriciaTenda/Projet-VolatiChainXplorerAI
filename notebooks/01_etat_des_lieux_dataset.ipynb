```python
from pathlib import Path

root_projet = Path("__file__").resolve().parents[1]
data_path = root_projet / "data" / "exports" / "datasets" / "btc_macro_dataset.csv"
print("Dossier courant :", Path.cwd())
print(data_path)
print("Fichier trouvé :", data_path.exists())
```

```python
import pandas as pd

df = pd.read_csv(data_path)

df.head(10)
```

```python
print("Dimensions :", df.shape)
print("\nColonnes :", df.columns.tolist())
print("\nTypes :\n", df.dtypes)
print("\nAperçu :\n", df.head())
print("\nValeurs manquantes :\n", df.isna().sum())
print("\nNombre de doublons :", df.duplicated().sum())
print("\nStatistiques :\n", df.describe())
```

```python
# Conversion de la date
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Contrôles temporels
print("Dates invalides :", df["date"].isna().sum())
print("Première date :", df["date"].min())
print("Dernière date :", df["date"].max())
print("Dates triées :", df["date"].is_monotonic_increasing)
print("Dates dupliquées :", df["date"].duplicated().sum())

# Recherche des jours absents
dates_attendues = pd.date_range(
    start=df["date"].min(),
    end=df["date"].max(),
    freq="D"
)

dates_manquantes = dates_attendues.difference(df["date"])

print("Nombre de dates manquantes :", len(dates_manquantes))
print("Premières dates manquantes :", dates_manquantes[:10])

# Analyse des valeurs macroéconomiques manquantes
macro_columns = [
    "rate_mro",
    "inflation_rate",
    "unemployment_rate",
    "monetary_m3_rate"
]

lignes_macro_manquantes = df[df[macro_columns].isna().any(axis=1)]

print("Première date macro manquante :", lignes_macro_manquantes["date"].min())
print("Dernière date macro manquante :", lignes_macro_manquantes["date"].max())
display(lignes_macro_manquantes.head())
display(lignes_macro_manquantes.tail())

# Contrôle du volume
print("Volumes égaux à zéro :", (df["volume"] == 0).sum())
display(df.loc[df["volume"] == 0, ["date", "volume"]].head(10))
```

```python
# Observer les données autour des deux dates absentes
display(
    df.loc[
        df["date"].between("2022-06-25", "2022-07-02"),
        ["date", "open", "high", "low", "close", "volume", "market_cap"]
    ]
)

# Nombre de volumes nuls par année
volume_zero = df.loc[df["volume"] == 0, ["date", "volume"]].copy()
volume_zero["annee"] = volume_zero["date"].dt.year

display(
    volume_zero.groupby("annee")
    .size()
    .rename("nombre_volume_zero")
)

# Identifier les périodes consécutives de volume nul
volume_zero["groupe"] = (
    volume_zero["date"].diff().dt.days.ne(1).cumsum()
)

periodes_volume_zero = (
    volume_zero.groupby("groupe")
    .agg(
        debut=("date", "min"),
        fin=("date", "max"),
        nombre_jours=("date", "size")
    )
)

display(periodes_volume_zero)

# Vérifier la cohérence des prix OHLC
ohlc_invalides = df[
    (df["high"] < df[["open", "close", "low"]].max(axis=1))
    | (df["low"] > df[["open", "close", "high"]].min(axis=1))
    | (df[["open", "high", "low", "close"]] <= 0).any(axis=1)
]

print("Lignes OHLC incohérentes :", len(ohlc_invalides))
```

```python
print("Volumes négatifs :", (df["volume"] < 0).sum())
print(
    "Capitalisations nulles ou négatives :",
    (df["market_cap"] <= 0).sum()
)
```
