# Cadre théorique et méthodologique




## 1. Problématique financière

L’objectif de VolatiChainXplorerAI n’est pas de prédire directement le prix futur du Bitcoin, mais d’estimer son niveau de volatilité future.

La volatilité représente l’intensité des variations du marché. Sa prévision peut aider à évaluer le risque associé à une période future, indépendamment de la direction haussière ou baissière du prix.

Deux horizons sont étudiés :

7 jours, pour une estimation du risque à court terme ;
30 jours, pour une estimation du risque à moyen terme.

## 2. Traduction statistique du besoin

Le prix du Bitcoin évolue fortement au fil du temps et présente généralement une tendance non stationnaire. Comparer directement ses variations absolues entre différentes périodes peut donc être trompeur.

Le projet utilise les rendements logarithmiques :

$$ r_t=\ln\left(\frac{P_t}{P_{t-1}}\right) $$

Cette transformation mesure le mouvement relatif du prix et rend les observations appartenant à différentes périodes plus comparables.

La volatilité est ensuite mesurée par l’écart-type des rendements.

## 3. Définition des variables expliquées

Les variables expliquées, ou cibles \(y\), correspondent à la volatilité réellement observée après chaque date \(t\) :

target_volatility_7d : volatilité des rendements de \(t+1\) à \(t+7\) ;
target_volatility_30d : volatilité des rendements de \(t+1\) à \(t+30\).

Ces variables représentent ce que les modèles devront apprendre à prédire.

## 4. Définition des variables explicatives

Les variables explicatives \(X\) doivent être disponibles à la date \(t\). Elles représentent les informations susceptibles d’expliquer la volatilité future :

rendements passés ;
volatilités historiques à 3, 7, 14 et 30 jours ;
amplitude journalière ;
variations du volume et de la capitalisation ;
variables retardées des rendements ;
indicateurs macroéconomiques disponibles à la date observée.

Les prix et volumes bruts sont conservés pour construire et contrôler ces variables, mais leur utilisation directe par les modèles doit être justifiée par l’analyse exploratoire.

## 5. Prévention des fuites temporelles

Pour chaque ligne datée de \(t\) :

les variables \(X\) utilisent uniquement les informations connues au plus tard à \(t\) ;
la cible \(y\) utilise les observations postérieures à \(t\).

Les indicateurs macroéconomiques sont décalés pour tenir compte de leur délai de publication. Les données sont séparées chronologiquement et une **zone d’embargo **empêche les fenêtres futures des cibles de chevaucher les périodes suivantes.

## 6. Analyse exploratoire et sélection des variables

Avant de figer les variables du modèle, l’analyse doit examiner :

- les distributions et valeurs extrêmes ;
- les valeurs manquantes ;
- l’évolution temporelle ;
- la stationnarité ;
- les corrélations et redondances ;
- les relations entre les variables explicatives et chaque cible ;
- la stabilité de ces relations dans le temps.

Cette analyse permettra de confirmer, supprimer ou compléter les variables candidates avant l’entraînement.

## 7. Structure statistique finale

Chaque observation du dataset représente une date \(t\) :

$$ X_t \longrightarrow y_{t+h} $$

où :

\(X_t\) est le vecteur des informations disponibles à la date \(t\) ;
\(y_{t+h}\) est la volatilité réalisée pendant l’horizon futur ;
\(h\) vaut 7 ou 30 jours.

Deux datasets de modélisation sont donc construits à partir d’une même base, chacun possédant sa propre cible.
